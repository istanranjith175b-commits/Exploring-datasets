```python
import matplotlib.pyplot as plt
pie_data = df['Department'].value_counts()
labels = pie_data.index
sizes = pie_data.values

plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Distribution of Departments")
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Assuming you want to plot the distribution of 'Department'
bar_data = df['Department'].value_counts()
categories = bar_data.index
values = bar_data.values

plt.figure(figsize=(10, 6))
plt.bar(categories, values, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.title('Number of Students per Department')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Plotting a histogram for 'Study_Hours'
plt.figure(figsize=(10, 6))
plt.hist(df['Study_Hours'], bins=5, edgecolor='black', color='skyblue')
plt.xlabel('Study Hours')
plt.ylabel('Frequency')
plt.title('Distribution of Study Hours')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the correlation matrix for numerical columns
correlation_matrix = df.select_dtypes(include=['number']).corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
from pandas.plotting import parallel_coordinates
```

```python
numerical_cols = ['Study_Hours', 'Attendance', 'Maths', 'Physics', 'Programming', 'English', 'Assignments', 'Final_Score']
df_plot = df[numerical_cols + ['Department']]

plt.figure(figsize=(12, 8))
parallel_coordinates(df_plot, 'Department', colormap='viridis')
plt.title('Parallel Coordinates Plot of Student Data by Department')
plt.xlabel('Metrics')
plt.ylabel('Value')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
from pandas.plotting import andrews_curves
```

```python
numerical_cols = ['Study_Hours', 'Attendance', 'Maths', 'Physics', 'Programming', 'English', 'Assignments', 'Final_Score']
df_andrews_plot = df[numerical_cols + ['Department']]

plt.figure(figsize=(12, 8))
andrews_curves(df_andrews_plot, 'Department', colormap='jet')
plt.title('Andrews Curves Plot of Student Data by Department')
plt.grid(True)
plt.show()
```
