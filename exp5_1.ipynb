```python
import pandas as pd
import numpy as np
data = {'weight': [60, 65, 70, 75, 80, 85, 90, 95, 100, 105],
        'bins': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C']}
df = pd.DataFrame(data)
display(df)
bins = [50, 70, 90, 110]
labels = ['Light', 'Medium', 'Heavy']

df['weight_category'] = pd.cut(df['weight'], bins=bins, labels=labels)
display(df)
pd.value_counts(df['weight_category'])
```
